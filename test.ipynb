```python
from annotate import run_pipeline

chat = """
Therapist: Hi, it’s good to see you today. What would you like to focus on?

Patient: I think… I need help quitting smoking. I’ve tried before, but it never lasts.

Therapist: I’m glad you brought this up. Quitting can be challenging, and many people need multiple attempts before it sticks. What makes you want to try again now?

Patient: I get breathless easily these days. And my family keeps telling me to stop. I guess I’m also worried about long-term health.

Therapist: Those sound like important reasons—your health and your family. On a scale from 0 to 10, how ready do you feel to quit?

Patient: Maybe a 6.

Therapist: A 6 is a solid starting point. Why not a lower number, like 3 or 4?

Patient: Because I do want to quit. I’m just not confident I can.

Therapist: That makes sense. Wanting to quit and feeling unsure can exist together. Let’s explore what makes it hard. When do you usually feel the urge to smoke?

Patient: Mostly after meals, with tea, and when I’m stressed. Especially at work.

Therapist: So smoking is tied to routines and stress relief. What do you feel smoking gives you in those moments?

Patient: It calms me down. Feels like a break.

Therapist: That’s important to recognize. We’re not just removing smoking—we’re replacing what it provides. Would you be open to trying alternative ways to manage those moments?

Patient: Like what?

Therapist: For example, after meals, you could try a short walk or chewing something like fennel seeds or gum. For stress, maybe deep breathing, or even a 2-minute pause away from your desk. What do you think might work for you?

Patient: Maybe the walking part. I can try that.

Therapist: That’s a great start. Small, practical changes often work best. Have you ever set a quit date before?

Patient: Yes, but I usually relapse within a week.

Therapist: That’s actually very common. Relapse isn’t failure—it’s part of the learning process. This time, we can plan for those difficult moments. Would you like to set a quit date again?

Patient: Maybe… next Monday?

Therapist: That sounds reasonable. Between now and then, we can prepare. One helpful step is reducing the number of cigarettes gradually. Another is identifying triggers and planning responses, like the walking idea you mentioned.

Patient: Okay, I can try cutting down.

Therapist: Also, some people benefit from nicotine replacement options like patches or gums. Have you considered those?

Patient: I’ve heard about them, but never tried.

Therapist: They can reduce withdrawal symptoms and make quitting more manageable. You could discuss this with a doctor or pharmacist.

Patient: I’ll think about it.

Therapist: Before we wrap up, let me summarize: You want to quit because of your health and family. You’re planning to cut down and set a quit date for next Monday. You’ll try walking after meals instead of smoking. Did I get that right?

Patient: Yes, that sounds right.

Therapist: That’s a strong plan. How confident do you feel about taking these steps?

Patient: Maybe a 7 now.

Therapist: That’s encouraging progress. We’ll keep building on this. When we meet next, we can review how things went and adjust if needed.

Patient: Okay. Thank you. I feel a bit more hopeful now.

Therapist: I’m glad to hear that. You’re taking meaningful steps, and that matters.
"""

row = run_pipeline(chat, conversation_id="demo-001")
print(row)
```
